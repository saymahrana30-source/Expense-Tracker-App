```python
raw = load_raw_data()
print('Shape:', raw.shape)
raw.head(8)
```

## 2. Clean the data

## 3. Add features and analyse

We add month, weekday and needs/wants columns, then summarise the spending.

```python
df = add_features(clean_df)
expenses = df[df['Type'] == 'Expense']
category_summary(expenses)
```

```python
monthly_summary(df)
```

```python
over_budget = budget_check(expenses)
over_budget[over_budget['Over_Budget']]
```

```python
detect_anomalies(expenses)[['Date', 'Category', 'Description', 'Amount']]
```

## 4. Draw the charts

Each function saves a PNG into `images/`. Display the two charts below.

```python
summary = category_summary(expenses)
monthly = monthly_summary(df)
for path in [plot_category_totals(summary), plot_monthly_income_expense(monthly)]:
    display(Image(filename=str(path), width=700))
```
